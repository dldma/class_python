## 2주차

```python
# 도미와 빙어를 같은 산점도에 표시하여
# 두 생선의 길이와 무게 분포를 비교한다.

plt.scatter(bream_length, bream_weight)
plt.scatter(smelt_length, smelt_weight)

plt.xlabel("Length")
plt.ylabel("Weight")

# [코드 전용 ZIP] images 폴더가 없으므로 기존 이미지 저장 명령을 주석 처리함.
# plt.savefig("../../images/offline_week01_knn_bream_smelt_scatter.png")

plt.show()
# [오류 수정] 오프라인 그래프를 ZIP의 images 폴더에 온라인 파일과 구분하여 저장하도록 수정했습니다.
```

```python
# 도미와 빙어의 길이 데이터를 하나의 리스트로 합친다.
# 도미와 빙어의 무게 데이터를 하나의 리스트로 합친다.

length = bream_length + smelt_length
weight = bream_weight + smelt_weight

print (length), print (weight)
print (len(length))
print (len(weight))
```

```python
fish_data = [[L,W] for L, W in zip(length, weight)]
print (fish_data)

fish_target = [1]*35 + [0]*14
print(fish_target)

# ─── 중간고사 학습 주석 (원본 코드 수정 없음) ───
# zip: 같은 위치에 있는 길이·무게를 짝지어 입력 특성 쌍을 만든다.
```

```python
from sklearn.neighbors import KNeighborsClassifier

#뒤에 ()가 붙는 다는 것은 이 이름의 함수라는 것이다
kn = KNeighborsClassifier()


kn.fit(fish_data, fish_target) #학습
kn.score(fish_data, fish_target) #시험

# ─── 중간고사 학습 주석 (원본 코드 수정 없음) ───
# KNeighborsClassifier: 가까운 k개 이웃의 클래스로 분류하는 모델이며 기본 n_neighbors는 5다.
# fit(X, y): 입력 데이터와 정답을 바탕으로 모델을 훈련한다.
# score(X, y): 분류에서는 정확도, 회귀에서는 결정계수 R²를 반환한다.
```

```python
#생선 분류
one = [[30,600]] #이 함수는 2차원 배열로 넣어야 함
kn.predict(one)     # 


# 학습데이터의 전체에 대해 분류 결과확인
pred = kn.predict(fish_data)
proba = kn.predict_proba(fish_data)
print(pred)
# print(proba)

# 분류결과와 확률확인

print(kn.predict(one),kn.predict_proba(one))


# ─── 중간고사 학습 주석 (원본 코드 수정 없음) ───
# predict_proba(X): 클래스별 예측 확률을 반환한다. 열 순서는 모델의 classes_ 순서다.
# predict(X): 학습된 모델이 입력 샘플의 클래스 또는 연속값을 예측한다.
```

### 온라인 1주차에서 보충 — 새 생선의 예측 결과를 산점도에서 확인

```python
# [온라인 1주차 보충] 원래 오프라인 코드에 없던 새 샘플의 위치 표시
# ① 도미와 빙어 데이터의 분포를 확인한다.
plt.scatter(bream_length, bream_weight, label='bream (1)')
plt.scatter(smelt_length, smelt_weight, label='smelt (0)')

# ② 기존 오프라인에서 사용한 one = [[30, 600]]을 그대로 사용한다.
# one[0][0] = 길이(30), one[0][1] = 무게(600)
plt.scatter(one[0][0], one[0][1], marker='^', s=120, label='new fish')
plt.xlabel('length')
plt.ylabel('weight')
plt.legend()
plt.show()

# ③ predict()는 클래스 0(빙어) 또는 1(도미)을 돌려준다.
print('new fish prediction:', kn.predict(one))
```
