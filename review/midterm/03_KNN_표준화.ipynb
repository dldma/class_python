### 온라인 3주차에서 보충 — `np.vstack()`, `np.full()`

```python
# [온라인 3주차 보충] 오프라인에서는 다루지 않은 NumPy 배열 함수 2개
# np.vstack(): 여러 배열을 세로(행) 방향으로 쌓는다.
row_stacked = np.vstack(([1, 2, 3], [4, 5, 6]))
print('vstack result:\n', row_stacked)  # 2행 3열

# np.full((행, 열), 값): 지정한 크기의 배열을 같은 값으로 채운다.
filled = np.full((2, 3), 9)
print('full result:\n', filled)  # 2행 3열, 모든 원소가 9

# 비교: 기존 오프라인 np.column_stack()은 원소를 열 단위로 배치한다.
print('column_stack result:\n', np.column_stack(([1, 2, 3], [4, 5, 6])))
```

```python
#훈련데이터와 타깃데이터를 나눈다
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target = train_test_split(
    fish_data,
    fish_target,
    random_state=42
)

print(train_input.shape, test_input.shape)
print(train_target.shape, test_target.shape)
print(train_target)
print(test_target)

# ─── 중간고사 학습 주석 (원본 코드 수정 없음) ───
# train_test_split(X, y, ...): 입력 X와 정답 y를 같은 인덱스로 분할해 훈련·테스트 세트를 만든다.
# random_state: 데이터 분할의 난수 상태를 고정해 반복 실행 시 같은 분할을 재현한다.
# shape: NumPy 배열의 차원별 크기를 보여준다. 예: (42, 1)은 42행 1열.
```

### 온라인 3주차에서 보충 — `stratify`로 클래스 비율 유지

```python
# [온라인 3주차 보충] 분류의 원본 정답(도미=1, 빙어=0) 비율을 고려해 분할한다.
# stratify=fish_target: train/test 세트의 클래스 비율이 전체와 비슷하도록 나눈다.
# 기존 오프라인 예제의 train_input 등을 덮어쓰지 않도록 별도 이름을 사용한다.
(
    strat_train_input, strat_test_input,
    strat_train_target, strat_test_target
) = train_test_split(
    fish_data, fish_target,
    stratify=fish_target,
    random_state=42
)

# 정답이 0과 1이므로 평균은 도미(1) 비율을 뜻한다.
print('전체 도미 비율:', np.mean(fish_target))
print('훈련 도미 비율:', np.mean(strat_train_target))
print('테스트 도미 비율:', np.mean(strat_test_target))
print('분할 후 입력 배열:', strat_train_input.shape, strat_test_input.shape)
```

```python
# 최근접 이웃
# 2시 34분 사진에 값넣기

from sklearn.neighbors import KNeighborsClassifier
kn = KNeighborsClassifier()

# 훈련
kn.fit(train_input, train_target)

print(kn.score(train_input, train_target)) #훈련데이터의 점수
print(kn.score(test_input, test_target))    #테스트 데이터의 점수


# ─── 중간고사 학습 주석 (원본 코드 수정 없음) ───
# KNeighborsClassifier: 가까운 k개 이웃의 클래스로 분류하는 모델이며 기본 n_neighbors는 5다.
# fit(X, y): 입력 데이터와 정답을 바탕으로 모델을 훈련한다.
# score(X, y): 분류에서는 정확도, 회귀에서는 결정계수 R²를 반환한다.
```

```python
#샘플평가하기
one = np.array([[25,150]])
distances, idx = kn.kneighbors(one)
print(kn.predict(one), kn.predict_proba(one))
kn.kneighbors(one)  #one과의 이웃으 정보(거리, 인덱스)

# ─── 중간고사 학습 주석 (원본 코드 수정 없음) ───
# predict_proba(X): 클래스별 예측 확률을 반환한다. 열 순서는 모델의 classes_ 순서다.
# predict(X): 학습된 모델이 입력 샘플의 클래스 또는 연속값을 예측한다.
# kneighbors(X): 가까운 이웃까지의 거리와 훈련 세트 내 인덱스를 돌려준다.
# np.array: 파이썬 리스트 등을 NumPy 배열로 변환한다.
```
