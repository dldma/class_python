```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# 1. 10개의 1로 이루어진 1차원 배열 arr1 생성
arr1 = np.ones(10)
print(arr1)
```

```python
# 2. 10개의 0으로 이루어진 1차원 배열 arr2 생성
arr2 = np.zeros(10)
print(arr2)
```

```python
# 3. arr1과 arr2를 연결하여 arr3 생성
arr3 = np.concatenate((arr1, arr2))
print(arr3)
```

```python
# 4. 3으로 채워진 3행 5열 배열 arr4 생성
arr4 = np.full((3,5),3)
print(arr4)
```

```python
# 5. arr1과 arr2를 열방향으로 연결하여 arr5 생성
arr5 = np.column_stack((arr1,arr2))
print(arr5)
```

```python
# 6. arr1과 arr2를 행방향으로 연결하여 arr6 생성
arr6 = np.row_stack((arr1,arr2))
print(arr6)
```

```python
# 7. 데이터를 훈련 데이터와 테스트 데이터로 나누기
# 유도선수는 1, 경마선수는 0
data = [[180,80],[175,70],[190,88],[190,80],[187,92],[160,54],[165,52],[155,49],[153,43]]
target = [1, 1, 1, 1, 1, 0, 0, 0, 0]

data = np.array([[180,80],[175,70],[190,88],[190,80],[187,92],[160,54],[165,52],[155,49],[153,43]])
target = np.array([1, 1, 1, 1, 1, 0, 0, 0, 0])

train_data, test_data, train_target, test_target = train_test_split(
    data,
    target,
    test_size=0.2,
    random_state=42
)

print("훈련 데이터")
print(train_data)

print("훈련 데이터 정답")
print(train_target)

print("테스트 데이터")
print(test_data)

print("테스트 데이터 정답")
print(test_target)
```

```python
# 8. data를 산점도로 표현
plt.scatter(
    data[:, 0], # 모든 선수의 키
    data[:, 1],   #모든 선수의 몸무게
    c=target
)

plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Original Data")
plt.show()
```

```python
# 9. data를 표준점수로 변환하여 scaled_data에 저장
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
print(scaled_data)
```

```python
# 10. scaled_data를 산점도로 표현
plt.scatter(
    scaled_data[:, 0],
    scaled_data[:, 1],
    c=target
)

plt.xlabel("Scaled Height")
plt.ylabel("Scaled Weight")
plt.title("Scaled Data")
plt.show()
```
