# Week 01 - K-최급접 이웃

도미와 빙어의 길이와 무게 데이터를 이용하여 K-최근접 이울 알고리즘으로 생성을 분류하는 실습

```python
# 도미의 길이와 무게 데이터를 각각 리스트로 저장한다.
# 같은 위치의 길이와 무게 값이 한 마리의 도미 데이터이다.

bream_length = [
    25.4, 26.3, 26.5, 29.0, 29.0, 29.7, 29.7,
    30.0, 30.0, 30.7, 31.0, 31.0, 31.5, 32.0,
    32.0, 32.0, 33.0, 33.0, 33.5, 33.5, 34.0,
    34.0, 34.5, 35.0, 35.0, 35.0, 35.0, 36.0,
    36.0, 37.0, 38.5, 38.5, 39.5, 41.0, 41.0
]

bream_weight = [
    242.0, 290.0, 340.0, 363.0, 430.0, 450.0, 500.0,
    390.0, 450.0, 500.0, 475.0, 500.0, 500.0, 340.0,
    600.0, 600.0, 700.0, 700.0, 610.0, 650.0, 575.0,
    685.0, 620.0, 680.0, 700.0, 725.0, 720.0, 714.0,
    850.0, 1000.0, 920.0, 955.0, 925.0, 975.0, 950.0
]
```

```python
# matplotlib pyplot을 plt라는 별칭으로 불러온다.
# 도미의 길이를 x축, 무게를 y축으로 하여 산점도를 그린다.

import matplotlib.pyplot as plt

plt.scatter(bream_length, bream_weight)
plt.xlabel("Length")
plt.ylabel("Weight")

plt.savefig("images/week01_knn_bream_scatter.png")
#수업외 결과를 이미지로 저장하는 함수
#한번 저장하고 다시 실행하면 기존 이미지에 덮어씨움

plt.show()
```

```python
#빙어 14마리의 길이와 무게 데이터를 저장한다
#같은 위치의 길이와 무게 값이 한 마리의 빙어 데이터이다.

smelt_length = [9.8, 10.5, 10.6, 11.0, 11.2, 11.3, 11.8,
    11.8, 12.0, 12.2, 12.4, 13.0, 14.3, 15.0
]

smelt_weight = [
    6.7, 7.5, 7.0, 9.7, 9.8, 8.7, 10.0,
    9.9, 9.8, 12.2, 13.4, 12.2, 19.7, 19.9
]
```

```python
# 도미와 빙어를 같은 산점도에 표시하여
# 두 생선의 길이와 무게 분포를 비교한다.

plt.scatter(bream_length, bream_weight)
plt.scatter(smelt_length, smelt_weight)

plt.xlabel("Length")
plt.ylabel("Weight")

plt.savefig("images/week01_knn_bream_smelt_scatter.png")

plt.show()
```

```python
# 도미와 빙어의 길이 데이터를 하나의 리스트로 합친다.
# 도미와 빙어의 무게 데이터를 하나의 리스트로 합친다.

length = bream_length + smelt_length
weight = bream_weight + smelt_weight
# 여기서 +연산자는 두 리스트를 이어 붙이는 역할을 한다.

print (length)
print (len(length))
print (weight)
print (len(weight))
```

```python
# zip()으로 길리와 무게를 하나씩 가져와서 하나의 튜플로 묶어준다.
# 튜플로 묶는다 => 각 생선을 [길이, 무게] 형태의 2차원 데이터로 만든다. 

fish_data = [[l,w] for l, w in zip(length,weight)]

print(fish_data)
print(len(fish_data))
```

```python
# 도미는 1, 빙어는 0으로 표시한 정답데이터를 만든다. 
# 앞에 35개는 도미, 뒤에 14개가 빙어이다

fish_target = [1]*35 + [0]*14

print(fish_target)
print(len(fish_target))
```

```python
# scikit-learn에서 k-최근접 이웃 분류기를 불러온다.
# 기본값으로 가장 가까운 5개의 데이터를 기준으로 분류한다.

from sklearn.neighbors import KNeighborsClassifier

kn = KNeighborsClassifier()
```

```python
# fish_data와 fish_target을 이용해
# k-최근접 이웃 모델을 학습시킨다.

kn.fit(fish_data, fish_target)
```

```python
# 학습한 모델이 현재 데이터를 얼마나 정확하게 분류하는지 확인한다.

kn.score(fish_data, fish_target)
```

```python
# 길이 30cm, 무게 600g인 생선을 입력해
# 도미인지 빙어인지 예측한다.

kn.predict([[30, 600]])
# [[]] 두겹인 이유는 지금은 하나의 데이터만 입력했지만
# 나중에 여러 개의 데이터를 한 번에 입력할 수 있도록 2차원 배열 형태로 입력해야 하기 때문이다.
```

```python
# 기존 도미와 빙어 데이터에
# 새로 예측한 생선 [30, 600]을 삼각형으로 표시한다.

plt.scatter(bream_length, bream_weight)
plt.scatter(smelt_length, smelt_weight)
plt.scatter(30, 600, marker='^')

plt.xlabel("Length")
plt.ylabel("Weight")

plt.savefig("images/week01_knn_prediction_scatter.png")

plt.show()
```

```python
# 모든 49개의 데이터를 이웃으로 사용하도록
# k-최근접 이웃 분류기를 새로 만든다.

kn49 = KNeighborsClassifier(n_neighbors=49)

# 동일한 fish_data와 fish_target으로
# 49개의 이웃을 사용하는 모델을 학습시킨다.

kn49.fit(fish_data, fish_target)
```

```python
# 49개의 데이터를 모두 이웃으로 사용하는 모델의
# 분류 정확도를 확인한다.

print(kn49.score(fish_data, fish_target))

# 전체 49마리 중 도미가 35마리이므로
# 항상 도미라고 판단했을 때의 비율을 계산한다.

print(35 / 49)

# n_neighbors를 과반 수 이상으로 설정하면
# 모델이 항상 도미라고 판단하게 되어
# 분류 정확도가 높게 나오는 것처럼 보이지만
# 무조건 도미라고 판단하는 모델이므로 빙어를 구분하지 못한다.
```

## 실습 정리

- 도미 35마리와 빙어 14마리의 길이와 무게 데이터를 준비했다.
- `matplotlib`의 `scatter()`를 이용해 데이터를 시각화했다.
- `zip()`과 리스트 내포를 이용해 `[길이, 무게]` 형태의 데이터를 만들었다.
- 도미는 `1`, 빙어는 `0`으로 target 데이터를 만들었다.
- `KNeighborsClassifier`를 이용해 k-최근접 이웃 모델을 생성했다.
- `fit()`으로 모델을 학습했다.
- `score()`로 모델의 성능을 확인했다.
- `predict()`로 새로운 생선을 예측했다.
- `n_neighbors` 값을 바꾸면 분류 결과와 정확도가 달라질 수 있음을 확인했다.
