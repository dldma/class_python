```python
# matplotlib pyplot을 plt라는 별칭으로 불러온다.
# 도미의 길이를 x축, 무게를 y축으로 하여 산점도를 그린다.

import matplotlib.pyplot as plt

plt.scatter(bream_length, bream_weight)
plt.xlabel("Length")
plt.ylabel("Weight")

plt.show()
```

```python
# 도미와 빙어의 길이 데이터를 하나의 리스트로 합친다.
# 도미와 빙어의 무게 데이터를 하나의 리스트로 합친다.

length = bream_length + smelt_length
weight = bream_weight + smelt_weight

print (length), print (weight)
print (len(length))
print (len(weight))
```

```python
fish_data = [[L,W] for L, W in zip(length, weight)]
print (fish_data)

fish_target = [1]*35 + [0]*14
print(fish_target)
```

```python
# 훈련데이터와 테스트데이터를 나누어 학습한다
train_input = fish_data[:35] # 훈련데이터 (0-49)
train_target = fish_target[:35]
test_input = fish_data[35:]
test_target = fish_target[35:]
print(len(train_input), len(train_target))
print(len(test_input), len(test_target))
```

```python
# 학습하고 점수 확인
# 문제점 : 학습데이터에는 도미만, 테스트 데이터는 빙어만
# 공부는 도미로 했지만 테스트는 빙어로 한 것

from sklearn.neighbors import KNeighborsClassifier

kn = KNeighborsClassifier()
kn = kn.fit(train_input, train_target)

kn.score(test_input, test_target)
```
