## 2주차

```python
# 도미와 빙어를 같은 산점도에 표시하여
# 두 생선의 길이와 무게 분포를 비교한다.

plt.scatter(bream_length, bream_weight)
plt.scatter(smelt_length, smelt_weight)

plt.xlabel("Length")
plt.ylabel("Weight")

plt.savefig("images/week01_knn_bream_smelt_scatter.png")

plt.show()
```

```python
# 도미와 빙어의 길이 데이터를 하나의 리스트로 합친다.
# 도미와 빙어의 무게 데이터를 하나의 리스트로 합친다.

length = bream_length + smelt_length
weight = bream_weight + smelt_weight

print (length), print (weight)
print (len(length))
print (len(weight))
```

```python
fish_data = [[L,W] for L, W in zip(length, weight)]
print (fish_data)

fish_target = [1]*35 + [0]*14
print(fish_target)
```

```python
from sklearn.neighbors import KNeighborsClassifier

#뒤에 ()가 붙는 다는 것은 이 이름의 함수라는 것이다
kn = KNeighborsClassifier()


kn.fit(fish_data, fish_target) #학습
kn.score(fish_data, fish_target) #시험
```

```python
#생선 분류
one = [[30,600]] #이 한수는 2차원 배열로 넣어야 함
kn.predict(one) 


# 학습데이터의 전체에 대해 분류 결과확인
pred = kn.predict(fish_data)
proba = kn.predict_proba(fish_data)
print(pred)
# print(proba)

# 분류결과와 확률확인

print(kn.predict(one),kn.predict_proba(one))

```
